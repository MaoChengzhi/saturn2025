# Juno quick event check

Same check and panels as the Cassini notebook (`cassini_quick_check.ipynb`), with `mission="juno"`.
The shared code is `wave_plots.quick_event_check`; Juno-specific reading is `jupiter/juno_data.py`,
geometry `jupiter/juno_spice.py`, and settings `missions.py` (`JUNO`).

**Options**
- `ion`: ICW ion, one of `H+`, `He+`, `O++`, `O+/S++`, `S+` (default), `SO+`, `SO2+`. The ICW band is 0.5–1.2 × its local gyrofrequency.
- `resolution`: `"sec"` (1 s daily files), `"full"` (native daily files), `"flyby"` (64 Hz moon-flyby files), `"auto"` (1 s if it resolves the ion, else native, else flyby).
- The field reference is `"local"`: Jupiter's tilted dipole rules out the Saturn spin-aligned mapping.
- `reader_options={"background_s": 300}` changes the FAC background window (default 1,200 s).
- The FFT panel marks the local gyrofrequencies of He+, O++, O+/S++, S+, SO+ and SO2+.

Juno times are printed to 1 ms, so 64 Hz samples are put back on their exact 15.625 ms grid inside each continuous run (largest shift 0.5 ms; field values unchanged).
