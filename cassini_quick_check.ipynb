## Documented events from the literature

Run the quick-check import above before selecting an event. All times are UTC. ### ICW · 2010-12-21 03:00–03:10 UTC

The authors use this equatorial interval to demonstrate ICW identification. They find a sharp spectral peak near the local water-ion gyrofrequency and check it against the transverse peak-to-peak periods. [Meeks et al. (2016), Fig. 3](https://doi.org/10.1016/j.pss.2016.06.003)

```python
ev_meeks_2010 = quick_event_check("2010-12-21 03:00:00", "2010-12-21 03:10:00", mission="cassini", ion="H2O+", resolution="auto")
```

### ICW inside 6 Saturn radii · 2006-07-24 00:50–00:53 UTC

At 4.7 Saturn radii and a mean background field of about 200 nT, they report left-hand circular polarization and a wave frequency near 0.17 Hz. [Chou & Cheng (2017), Fig. 3](https://doi.org/10.1186/s40623-017-0709-0)

```python
ev_chou_inner = quick_event_check("2006-07-24 00:46:30", "2006-07-24 00:56:30", mission="cassini", ion="H2O+", resolution="auto")
```

### ICW outside 6 Saturn radii · 2007-09-30 02:45–02:48 UTC

At 7.1 Saturn radii and a mean background field of about 50 nT, they report left-hand circular polarization at about 0.039 Hz. [Chou & Cheng (2017), Fig. 4](https://doi.org/10.1186/s40623-017-0709-0)

```python
ev_chou_outer = quick_event_check("2007-09-30 02:41:30", "2007-09-30 02:51:30", mission="cassini", ion="H2O+", resolution="auto")
```
