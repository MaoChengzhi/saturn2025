```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def cyclotron_frequency(B,m=18,q=1):
    return B*q*1.6e-19/(m*1.67e-27)
```

```python
# cyclotron freq of H2O+ at different magnetic field strengths
B=np.linspace(10e-9, 1000e-9, 100)


fig,ax=plt.subplots()
ax.plot(B*1e9, cyclotron_frequency(B))
ax.set_xlabel('Magnetic Field (nT)')
ax.set_ylabel('Cyclotron Frequency (Hz)')
ax.set_title('H2O+ Cyclotron Frequency vs Magnetic Field')
plt.show()
```

```python
# basic model of jupiter and saturn magnetosphere
r=np.linspace(1, 10, 100)
B_saturn= 0.21e-4*(1./r)**3
B_jupiter= 4.2e-4*(1./r)**3
```

```python
fig,ax=plt.subplots()
ax.plot(r, B_saturn, label='Saturn')
ax.plot(r, B_jupiter, label='Jupiter')
ax.set_xlabel('Distance (RJ)')
ax.set_ylabel('Magnetic Field (T)')
ax.set_title('Magnetosphere of Jupiter and Saturn')
ax.set_yscale('log')
ax.legend()
plt.show()
```

```python
# plot the cyclotron frequency vs magnetic field for H2O+
fig,ax=plt.subplots()
ax.plot(r,cyclotron_frequency(B_saturn), label='Saturn')
ax.plot(r,cyclotron_frequency(B_jupiter), label='Jupiter')
ax.set_xlabel('Distance (R)')
ax.set_ylabel('Cyclotron Frequency (Hz)')
ax.set_yscale('log')
ax.set_title('H2O+ Cyclotron Frequency vs Distance')
ax.legend()
plt.show()
```
