```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def n(r):
    return 1/(1+r**2)**2

def B(r):
    return -r/(1+r**2)
```

```python
r=np.linspace(0, 5, 100)
fig,ax=plt.subplots()
ax.plot(r, n(r), label='n(r)')
ax.plot(r, B(r), label='B(r)')
ax.set_xlabel('r')
ax.set_ylabel('Function values')
ax.set_title('Plots of n(r) and B(r)')
ax.legend()
```
