```python

fig, ax = plt.subplots(figsize=(8, 6))

# --- Axis arrows (manual, not using ax spines) ---
ax.annotate('', xy=(5.5, 0), xytext=(-3.5, 0),
            arrowprops=dict(arrowstyle='->', lw=1.5, color='black'))
ax.annotate('', xy=(0, 4.5), xytext=(0, -0.5),
            arrowprops=dict(arrowstyle='->', lw=1.5, color='black'))
ax.text(5.6, -0.15, r'$x$', fontsize=16, ha='left', va='top')
ax.text(-0.15, 4.6, r'$y$', fontsize=16, ha='right', va='bottom')

# --- k vector along x ---
kx, ky = 3.5, 0.0
ax.annotate('', xy=(kx, ky), xytext=(0.15, 0.0),
            arrowprops=dict(arrowstyle='->', lw=2.5, color='#2166AC'))
ax.text(kx / 2, -0.35, r'$\mathbf{k}$', fontsize=18, color='#2166AC',
        ha='center', va='top', fontweight='bold')

# --- B0 vector = (B0x, B0y, 0) ---
B0x, B0y = 3.0, 2.0
ax.annotate('', xy=(B0x, B0y), xytext=(0.0, 0.0),
            arrowprops=dict(arrowstyle='->', lw=2.5, color='#B2182B'))
ax.text(B0x + 0.15, B0y + 0.1, r'$\mathbf{B}_0$', fontsize=18,
        color='#B2182B', ha='left', va='bottom', fontweight='bold')

# Angle arc for theta between k and B0
theta_B0 = np.degrees(np.arctan2(B0y, B0x))
arc = patches.Arc((0, 0), 1.6, 1.6, angle=0, theta1=0, theta2=theta_B0,
                   color='gray', lw=1.2, linestyle='--')
ax.add_patch(arc)
ax.text(1.1, 0.35, r'$\theta$', fontsize=14, color='gray', ha='center')

# --- B1 vector = (0, B1y, B1z) shown in y-direction only (2D projection) ---
# Place B1 at a point along x to suggest it's a perturbation
B1_origin_x, B1_origin_y = 0.0, 0.0
B1x_val,B1y_val = -1.2,2.2
ax.annotate('', xy=(B1_origin_x+B1x_val, B1_origin_y + B1y_val),
            xytext=(B1_origin_x, B1_origin_y ),
            arrowprops=dict(arrowstyle='->', lw=2.5, color='#1A9850'))
ax.text(B1_origin_x + 0.2, B1_origin_y + B1y_val / 2 + 0.3,
        r'$\mathbf{B}_1$', fontsize=18, color='#1A9850',
        ha='left', va='center', fontweight='bold')

# --- z direction indicator (out of page) ---
circle_x, circle_y = -0.4, -0.4
circle = plt.Circle((circle_x, circle_y), 0.18, fill=False,
                     edgecolor='gray', lw=1.2)
ax.add_patch(circle)
# dot in center = out of page
ax.plot(circle_x, circle_y, 'o', color='gray', markersize=3)
ax.text(circle_x + 0.3, circle_y, r'$z$', fontsize=13, color='gray',
        ha='left', va='center')

# --- Legend / text box ---
textstr = (r'$\mathbf{B}_0 = (B_{0x},\; B_{0y},\; 0)$' + '\n'
           + r'$\mathbf{B}_1 = (0,\; \widetilde{B}_{1y},\; \widetilde{B}_{1z})$' + '\n'
           + r'$\mathbf{k} = k\,\hat{x}$')
props = dict(boxstyle='round,pad=0.5', facecolor='#F5F5F5',
             edgecolor='gray', alpha=0.9)
ax.text(4.8, 3.8, textstr, fontsize=13, verticalalignment='top',
        bbox=props, linespacing=1.8)

# --- Formatting ---
ax.set_xlim(-1, 6.5)
ax.set_ylim(-1, 5)
ax.set_aspect('equal')
ax.axis('off')
ax.set_title('MHD Wave Geometry', fontsize=16, pad=15)
```
