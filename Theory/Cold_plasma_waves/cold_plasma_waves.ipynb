## 1. Derivation: from the cold-fluid equations to a 3×3 matrix for **E**

**Cold fluid momentum equation** for species $s$ (charge $q_s$, mass $m_s$, density $n_s$), linearised about $\mathbf v_{s0}=0$, $\mathbf E_0=0$, with all perturbations $\propto e^{i(\mathbf k\cdot\mathbf x-\omega t)}$:

$$-i\omega m_s\mathbf v_s=q_s\left(\mathbf E+\mathbf v_s\times B_0\hat z\right).$$

Define the **signed** cyclotron frequency $\omega_{cs}=q_sB_0/m_s$ (so $\omega_{ce}=-\Omega_e<0$, $\omega_{ci}=+\Omega_i>0$) and the plasma frequency $\omega_{ps}^2=n_sq_s^2/(\varepsilon_0m_s)$. Solving for $\mathbf v_s$:

$$v_{sz}=\frac{iq_s}{m_s\omega}E_z,\qquad
\mathbf v_{s\perp}=\frac{iq_s}{m_s\omega}\,\frac{\mathbf E_\perp-\dfrac{i\omega_{cs}}{\omega}\,\hat z\times\mathbf E_\perp}{1-\omega_{cs}^2/\omega^2}.$$

**Current and dielectric tensor.** $\mathbf J=\sum_s n_sq_s\mathbf v_s=\overleftrightarrow{\sigma}\cdot\mathbf E$; Ampère's law with $\partial_t\to-i\omega$ gives $\nabla\times\mathbf B=-i\omega\mu_0\varepsilon_0\,\overleftrightarrow{\varepsilon}\cdot\mathbf E$ with $\overleftrightarrow{\varepsilon}=\overleftrightarrow{I}+\dfrac{i}{\omega\varepsilon_0}\overleftrightarrow{\sigma}$:

$$\overleftrightarrow{\varepsilon}=\begin{pmatrix}S&-iD&0\\ iD&S&0\\ 0&0&P\end{pmatrix}\qquad\text{(Bellan Eq. 6.11)}$$

**Wave equation.** Faraday + Ampère: $\mathbf k\times(\mathbf k\times\mathbf E)+\dfrac{\omega^2}{c^2}\overleftrightarrow{\varepsilon}\cdot\mathbf E=0$. With the refractive index $\mathbf n=\mathbf kc/\omega$, $n_x=n\sin\theta$, $n_y=0$, $n_z=n\cos\theta$ ($\theta$ = angle between $\mathbf k$ and $\mathbf B_0$):

$$\boxed{\begin{pmatrix}
S-n^2\cos^2\theta & -iD & n^2\sin\theta\cos\theta\\
iD & S-n^2 & 0\\
n^2\sin\theta\cos\theta & 0 & P-n^2\sin^2\theta
\end{pmatrix}
\begin{pmatrix}E_x\\E_y\\E_z\end{pmatrix}=0}\qquad\text{(Bellan Eq. 6.20)}$$

## 2. Definitions of S, D, P, R, L

General (any number of species, signed $\omega_{cs}$; Bellan Eq. 6.12, Stix 1992 Eq. 1-19):

| symbol | name (Stix mnemonic) | definition |
|---|---|---|
| $R$ | Right-hand circular | $\displaystyle R=1-\sum_s\frac{\omega_{ps}^2}{\omega(\omega+\omega_{cs})}$ |
| $L$ | Left-hand circular | $\displaystyle L=1-\sum_s\frac{\omega_{ps}^2}{\omega(\omega-\omega_{cs})}$ |
| $S$ | **S**um | $\displaystyle S=\tfrac12(R+L)=1-\sum_s\frac{\omega_{ps}^2}{\omega^2-\omega_{cs}^2}$ |
| $D$ | **D**ifference | $\displaystyle D=\tfrac12(R-L)=\sum_s\frac{\omega_{cs}}{\omega}\frac{\omega_{ps}^2}{\omega^2-\omega_{cs}^2}$ |
| $P$ | **P**arallel (= unmagnetised) | $\displaystyle P=1-\sum_s\frac{\omega_{ps}^2}{\omega^2}$ |

Useful identity: $S^2-D^2=RL$.

**Electron + one singly charged ion.** Use the CMA variables
$$X\equiv\frac{\omega_{pe}^2}{\omega^2},\qquad Y\equiv\frac{\Omega_e}{\omega}=\frac{|\omega_{ce}|}{\omega},\qquad \mu\equiv\frac{m_i}{m_e},$$
so $\omega_{pi}^2/\omega^2=X/\mu$ and $\Omega_i/\omega=Y/\mu$ (quasi-neutral, $n_i=n_e$). Then

$$R=1-\frac{X}{1-Y}-\frac{X}{\mu+Y},\qquad
L=1-\frac{X}{1+Y}-\frac{X}{\mu-Y},\qquad
P=1-X-\frac{X}{\mu}.$$

Resonances ($n\to\infty$): $R=\infty$ at $\omega=\Omega_e$ ($Y=1$, electron cyclotron); $L=\infty$ at $\omega=\Omega_i$ ($Y=\mu$, ion cyclotron).

In this sign convention the **R** wave rotates with the electrons and the **L** wave with the ions (Bellan §6.2.1).

## 3. Determinant = 0 → dispersion relation

Setting the determinant of the boxed matrix to zero (Bellan Eq. 6.47–6.49):

$$A\,n^4-B\,n^2+C=0,\qquad
\begin{aligned}
A&=S\sin^2\theta+P\cos^2\theta\\
B&=RL\sin^2\theta+PS\,(1+\cos^2\theta)\\
C&=PRL
\end{aligned}
\qquad\Rightarrow\qquad
n^2_\pm=\frac{B\pm\sqrt{B^2-4AC}}{2A},$$

with $B^2-4AC=(RL-PS)^2\sin^4\theta+4P^2D^2\cos^2\theta\ge 0$ (Eq. 6.50), so $n^2$ is always real: a mode either propagates ($n^2>0$) or is evanescent ($n^2<0$). An equivalent compact form is

$$\tan^2\theta=-\frac{P\,(n^2-R)(n^2-L)}{(Sn^2-RL)(n^2-P)}.$$

**Limits**
- $\theta=0$ ($\mathbf k\parallel\mathbf B_0$): $P\,[(n^2-R)(n^2-L)]=0$ → **R wave** $n^2=R$, **L wave** $n^2=L$ (circularly polarised, $iE_x/E_y=\pm1$), and the electrostatic oscillation $P=0$.
- $\theta=\pi/2$ ($\mathbf k\perp\mathbf B_0$): $(n^2-P)(Sn^2-RL)=0$ → **O mode** $n^2=P$ ($\mathbf E\parallel\mathbf B_0$) and **X mode** $n^2=RL/S$ (elliptically polarised in the plane $\perp\mathbf B_0$, with a longitudinal $E_x$).

**Cutoffs** ($n\to0$): $C=0$ ⇔ $P=0$, $R=0$ or $L=0$.
**Resonances** ($n\to\infty$): $A=0$ ⇔ $\tan^2\theta_{res}=-P/S$; at $\theta=0$ these are $R=\infty$, $L=\infty$; at $\theta=\pi/2$ they are $S=0$ (upper and lower hybrid).

The next cell checks this expansion symbolically.

```python
import sympy as sp
S_, D_, P_, n2_, th_ = sp.symbols('S D P n2 theta', real=True)
s, c = sp.sin(th_), sp.cos(th_)
M = sp.Matrix([[S_ - n2_*c**2, -sp.I*D_, n2_*s*c],
               [sp.I*D_, S_ - n2_, 0],
               [n2_*s*c, 0, P_ - n2_*s**2]])
R_, L_ = S_ + D_, S_ - D_
A = S_*s**2 + P_*c**2
B = R_*L_*s**2 + P_*S_*(1 + c**2)
C = P_*R_*L_
print('det - (A n^4 - B n^2 + C) =', sp.simplify(sp.expand(M.det() - (A*n2_**2 - B*n2_ + C))))
print('B^2-4AC - [(RL-PS)^2 sin^4 + 4 P^2 D^2 cos^2] =',
      sp.simplify(sp.expand(B**2 - 4*A*C - ((R_*L_ - P_*S_)**2*s**4 + 4*P_**2*D_**2*c**2))))
```

## 5. ω–k diagrams for θ = 0 and θ = 90°

Schematic-style figure like the lecture notes, but computed from $n^2=R,L$ ($\theta=0$) and $n^2=P,\;RL/S$ ($\theta=90°$) with $k=n\omega/c$.

Parameters: $\Omega_e/\omega_{pe}=0.5$ (dense plasma, $\omega_{pe}>\Omega_e$, same ordering as the notes) and a **reduced mass ratio $\mu=4$** so that the ion branches are visible on a linear axis. Section 6 repeats this on log axes with the real proton mass ratio.

Notation: $\omega_R,\omega_L$ = R/L cutoffs, $\omega_{UH}$ (written $\omega_{HH}$ in the lecture notes) = upper hybrid, $\omega_{LH}$ = lower hybrid, $v_A$ = Alfvén speed.

```python
def omega_k_figure(Oe=0.5, mu=4.0, kmax=3.0, wmax=2.0, labels=True, fname=None):
    f = cp.characteristic_frequencies(Oe, mu)
    w = np.concatenate([np.linspace(1e-4, wmax * 1.6, 60000)])
    par = cp.omega_k_branches(Oe, mu, 0.0, w)
    per = cp.omega_k_branches(Oe, mu, np.pi / 2, w)

    fig, axs = plt.subplots(1, 2, figsize=(13, 6.6), sharey=True)
    kk = np.linspace(0, kmax, 10)
    col = dict(R='#1f5aa6', L='#c0392b', O='#2a9d8f', X='#8e44ad')
    for ax in axs:
        ax.plot(kk, kk, ':', color='k', lw=1)                 # omega = k c
        ax.set_xlim(0, kmax); ax.set_ylim(0, wmax)
        ax.set_xlabel(r'$kc/\omega_{pe}$', fontsize=13)
        ax.set_xticks(np.arange(0.5, kmax + 0.01, 0.5))
        for sp_ in ('top', 'right'):
            ax.spines[sp_].set_visible(False)
    axs[0].set_ylabel(r'$\omega/\omega_{pe}$', fontsize=13)

    # --- theta = 0
    ax = axs[0]
    for name, segs in par.items():
        for k, ww in segs:
            ax.plot(k, ww, color=col[name], lw=2)
    ax.plot(kk, f['vA'] * kk, ':', color='gray', lw=1)       # omega = k vA
    for val, lab in [(f['Oe'], r'$\Omega_e$'), (f['Oi'], r'$\Omega_i$')]:
        ax.axhline(val, ls=':', color='gray', lw=1)
        ax.text(kmax * 1.01, val, lab, va='center', fontsize=13)
    for val, lab in [(f['wR'], r'$\omega_R$'), (f['wL'], r'$\omega_L$')]:
        ax.text(0.03, val + 0.03, lab, ha='left', va='bottom', fontsize=13)
    ax.set_title(r'$\mathbf{k}\parallel\mathbf{B}_0$  ($\theta=0$)', fontsize=14)
    if labels:
        ax.text(0.2, 1.85, 'EM wave (R wave)', color=col['R'], fontsize=13)
        ax.text(1.6, f['wL'] + 0.45, 'EM wave (L wave)', color=col['L'], fontsize=13)
        ax.text(1.55, 0.30, 'electron cyclotron wave', fontsize=12, color=col['R'])
        ax.text(0.42, 0.28, 'whistler', fontsize=12, color=col['R'])
        ax.text(1.7, f['Oi'] + 0.03, 'ion cyclotron wave', fontsize=12, color=col['L'])
        ax.text(2.4, f['vA'] * 2.4 + 0.04, r'$\omega=kv_A$', fontsize=12, color='gray')
        ax.text(0.0, -0.09, 'Alfvén', fontsize=13)
        ax.text(1.25, 1.12, r'$\omega=kc$', fontsize=12)

    # --- theta = 90 deg
    ax = axs[1]
    for name, segs in per.items():
        for k, ww in segs:
            ax.plot(k, ww, color=col[name], lw=2)
    for val, lab in [(f['wUH'], r'$\omega_{UH}$'), (f['wLH'], r'$\omega_{LH}$')]:
        ax.axhline(val, ls=':', color='gray', lw=1)
        ax.text(kmax * 1.01, val, lab, va='center', fontsize=13)
    for val, lab in [(f['wR'], r'$\omega_R$'), (f['wp'], r'$\omega_{p}$'), (f['wL'], r'$\omega_L$')]:
        ax.text(0.03, val + 0.02, lab, ha='left', va='bottom', fontsize=13)
    ax.set_title(r'$\mathbf{k}\perp\mathbf{B}_0$  ($\theta=90^\circ$)', fontsize=14)
    if labels:
        ax.text(0.2, 1.85, 'EM wave (X mode)', color=col['X'], fontsize=13)
        ax.text(1.55, 1.45, 'O mode', color=col['O'], fontsize=13)
        ax.text(1.95, f['wUH'] - 0.13, 'X mode', color=col['X'], fontsize=13)
        ax.text(1.95, f['wUH'] + 0.04, 'upper hybrid', fontsize=12)
        ax.text(1.3, f['wLH'] - 0.11, 'X mode (magnetosonic)', color=col['X'], fontsize=13)
        ax.text(1.95, f['wLH'] + 0.04, 'lower hybrid', fontsize=12)
        ax.text(0.0, -0.09, 'Alfvén', fontsize=13)
        ax.text(0.8, 0.68, r'$\omega=kc$', fontsize=12)
        fig.suptitle(f'Dispersion relations for propagation parallel and perpendicular to B0   (Ωe/ωpe = {Oe},  mi/me = {mu:g})', fontsize=15)
    fig.tight_layout()
    if fname:
        fig.savefig(FIG / fname, dpi=160, bbox_inches='tight')
    return fig, f

fig, f = omega_k_figure(fname='omega_k_theta0_theta90.png')
print({k: round(float(v), 4) for k, v in f.items()})
```

**Reading the figure.**
- $\theta=0$: the R wave has an upper branch above the cutoff $\omega_R$ and a lower branch (Alfvén → whistler → electron cyclotron wave, resonance at $\Omega_e$). The L wave has an upper branch above $\omega_L$ and a lower branch (Alfvén → ion cyclotron wave, resonance at $\Omega_i$). Between $\Omega_e$ and $\omega_L$ (or $\omega_R$) a parallel wave cannot propagate.
- $\theta=90°$: the O mode is unaffected by $B_0$ (cutoff at $\omega_p\approx\omega_{pe}$). The X mode has three branches: above $\omega_R$; from $\omega_L$ up to the upper hybrid resonance $\omega_{UH}$; and from the compressional (fast) Alfvén / magnetosonic wave at low $\omega$ up to the lower hybrid resonance $\omega_{LH}$.

## 6. Same branches on log axes with the real mass ratio (H⁺, μ = 1836)

For space-plasma use the ion branches lie many decades below the electron ones, so log axes are needed. Saturn's and Jupiter's magnetospheres are often water-group or sulfur/oxygen dominated; change `mu` (e.g. `16*cp.MU_PROTON`) to explore.

```python
def omega_k_loglog(Oe=0.5, mu=cp.MU_PROTON, fname=None):
    f = cp.characteristic_frequencies(Oe, mu)
    w = np.geomspace(1e-6, 10, 200000)
    fig, axs = plt.subplots(1, 2, figsize=(13, 5.6), sharey=True)
    for ax, th, title in [(axs[0], 0.0, r'$\theta=0$'), (axs[1], np.pi/2, r'$\theta=90^\circ$')]:
        br = cp.omega_k_branches(Oe, mu, th, w)
        colors = dict(R='#1f5aa6', L='#c0392b', O='#2a9d8f', X='#8e44ad')
        for name, segs in br.items():
            for i, (k, ww) in enumerate(segs):
                ax.loglog(k, ww, color=colors[name], lw=2, label=name if i == 0 else None)
        kk = np.geomspace(1e-6, 1e4, 10)
        ax.loglog(kk, kk, 'k:', lw=1); ax.loglog(kk, f['vA'] * kk, ':', color='gray', lw=1)
        names = (['Oe', 'Oi', 'wR', 'wL'] if th == 0 else ['wUH', 'wLH', 'wR', 'wL', 'wp'])
        tex = dict(Oe=r'$\Omega_e$', Oi=r'$\Omega_i$', wR=r'$\omega_R$', wL=r'$\omega_L$',
                   wUH=r'$\omega_{UH}$', wLH=r'$\omega_{LH}$', wp=r'$\omega_p$')
        for j, n_ in enumerate(sorted(names, key=lambda q: f[q])):
            ax.axhline(f[n_], ls=':', color='gray', lw=0.8)
            ax.text(3e3 / 8**j, f[n_], tex[n_], va='bottom', fontsize=11)
        ax.set_xlim(1e-6, 1e4); ax.set_ylim(1e-6, 10)
        ax.set_xlabel(r'$kc/\omega_{pe}$'); ax.set_title(title); ax.legend(loc='upper left')
    axs[0].set_ylabel(r'$\omega/\omega_{pe}$')
    fig.suptitle(f'Cold-plasma dispersion relation (log–log,  Ωe/ωpe = {Oe},  mi/me = {mu:.0f})')
    fig.tight_layout()
    if fname:
        fig.savefig(FIG / fname, dpi=160, bbox_inches='tight')
    return fig

omega_k_loglog(fname='omega_k_loglog_mu1836.png');
```

## 7. CMA diagram

Axes follow CMA-Lab/CMA-Diagram: horizontal $X=\omega_{pe}^2/\omega^2$ (∝ density), vertical $Y=\Omega_e/\omega$ (∝ $B_0$). Bellan Fig. 6.2 uses $(\omega_{pe}^2+\omega_{pi}^2)/\omega^2$ and $\Omega_e^2/\omega^2$ on log axes; the topology is the same.

Bounding surfaces (Bellan §6.2.7), solved for $X(Y)$ for one ion species:
- cutoffs: $P=0$: $X=\dfrac{\mu}{\mu+1}$; $\;R=0$: $X=\dfrac{(1-Y)(\mu+Y)}{\mu+1}$; $\;L=0$: $X=\dfrac{(1+Y)(\mu-Y)}{\mu+1}$
- resonances: $R=\infty$: $Y=1$; $\;L=\infty$: $Y=\mu$; $\;S=0$: $X=\left[\dfrac{1}{1-Y^2}+\dfrac{\mu}{\mu^2-Y^2}\right]^{-1}$

For a **fixed plasma** (fixed $n_e$, $B_0$) changing $\omega$ moves the point along the ray $Y\propto\sqrt X$ (a 45° line on Bellan's log–log axes): high $\omega$ near the origin, low $\omega$ far away.

Each bounded region is found automatically from the signs of $(P,R,L,S)$ and the position relative to $Y=1$ and $Y=\mu$; a small $v_{ph}/c$ wave-normal sketch is drawn at an interior point of each region ($\mathbf B_0$ vertical, both roots scaled together).

```python
def find_regions(mu, xmax, ymax, nx=500, ny=500, logscale=False):
    if logscale:
        xs = np.geomspace(xmax[0], xmax[1], nx); ys = np.geomspace(ymax[0], ymax[1], ny)
    else:
        xs = np.linspace(1e-3, xmax, nx); ys = np.linspace(1e-3, ymax, ny)
    XX, YY = np.meshgrid(xs, ys)
    code_ = cp.region_signature(XX, YY, mu)
    regions = []
    for c_ in np.unique(code_):
        lab, nlab = ndimage.label(code_ == c_)
        for j in range(1, nlab + 1):
            mask = lab == j
            if mask.sum() < 0.002 * mask.size:      # skip slivers
                continue
            dist = ndimage.distance_transform_edt(np.pad(mask, 1))[1:-1, 1:-1]
            iy, ix = np.unravel_index(np.argmax(dist), dist.shape)
            regions.append(dict(code=c_, x=xs[ix], y=ys[iy], area=mask.sum()))
    # order: top-to-bottom, left-to-right
    regions.sort(key=lambda r: (-round(r['y'] / (ymax if not logscale else 1), 1), r['x']))
    for i, r in enumerate(regions, 1):
        r['id'] = i
    return regions


def draw_boundaries(ax, mu, xmax, ymax, logscale=False):
    Y = np.geomspace(1e-4, ymax if not logscale else ymax[1], 6000) if logscale else np.linspace(0, ymax, 6000)
    b = cp.cma_boundaries(Y, mu)
    sty = {'R=0': ('#8c6d1f', '-'), 'L=0': ('#2a9d8f', '-'), 'S=0': ('#e9c46a', '-'), 'P=0': ('#264653', '-')}
    for name, X in b.items():
        ax.plot(X, Y, sty[name][1], color=sty[name][0], lw=2, label=name)
    ax.axhline(1, color='#3a86ff', lw=2, label=r'R=$\infty$ ($\omega=\Omega_e$)')
    ax.axhline(mu, color='#b5179e', lw=2, label=r'L=$\infty$ ($\omega=\Omega_i$)')


def wave_normal_sketch(ax, x, y, mu, size=0.055, labels=True):
    p = cp.stix(x, y, mu)
    th = np.linspace(0, 2*np.pi, 721)
    vp, vm = cp.phase_velocity(th, p['S'], p['P'], p['R'], p['L'])
    vmax = np.nanmax(np.concatenate([vp, vm]))
    fx, fy = ax.transLimits.transform(ax.transScale.transform((x, y)))
    names = cp.label_modes(p['S'], p['P'], p['R'], p['L'])
    for v, colr, nm in [(vp, '#2a9d8f', names[0]), (vm, '#c9a227', names[1])]:
        if not np.isfinite(v).any():
            continue
        r = v / vmax * size
        ax.plot(fx + r*np.sin(th), fy + r*np.cos(th), color=colr, lw=1.6, transform=ax.transAxes)
        if labels:
            dx = -0.008 if colr == '#2a9d8f' else 0.008      # keep the two roots' labels apart
            if nm[0]:
                ax.text(fx + dx, fy + np.nanmax(r[:3]) + 0.008, nm[0], fontsize=7, color='#50258c',
                        ha='center', transform=ax.transAxes)
            if nm[1]:
                ax.text(fx + np.nanmax(r[178:183]) + 0.008, fy + 1.5 * dx, nm[1], fontsize=7, color='#50258c',
                        va='center', transform=ax.transAxes)


def cma_figure(mu=2.5, xmax=1.5, ymax=3.5, fname=None):
    fig, ax = plt.subplots(figsize=(10, 8.5))
    ax.set_xlim(0, xmax); ax.set_ylim(0, ymax)
    draw_boundaries(ax, mu, xmax, ymax)
    regions = find_regions(mu, xmax, ymax)
    fig.canvas.draw()
    for r in regions:
        wave_normal_sketch(ax, r['x'], r['y'], mu)
        p_ = cp.stix(r['x'], r['y'], mu)
        if not any(np.isfinite(v).any() for v in cp.phase_velocity(np.linspace(0, np.pi, 181), p_['S'], p_['P'], p_['R'], p_['L'])):
            ax.text(r['x'], r['y'] - 0.12, 'no propagating mode\n(all evanescent)', ha='center', va='top', fontsize=8, color='0.35')
        ax.text(r['x'], r['y'], str(r['id']), fontsize=8, ha='center', va='center',
                bbox=dict(boxstyle='circle,pad=0.15', fc='white', ec='0.5', lw=0.6))
    ax.set_xlabel('X = ωpe²/ω²   (∝ density)', fontsize=13)
    ax.set_ylabel('Y = Ωe/ω   (∝ B0)', fontsize=13)
    ax.set_title(f'CMA diagram (cold plasma, electrons + one ion species, mi/me = {mu:g})\n'
                 'Each region: sketch of the v_ph/c wave-normal surface (B0 points up)', fontsize=13)
    ax.legend(loc='upper right', fontsize=9, framealpha=0.95)
    fig.tight_layout()
    if fname:
        fig.savefig(FIG / fname, dpi=170, bbox_inches='tight')
    return fig, regions

MU_TEACH = 2.5   # reduced mass ratio as in the CMA-Lab default and many textbooks, so all regions are visible
fig, regions = cma_figure(MU_TEACH, fname='cma_diagram_mu2p5.png')
print(len(regions), 'regions found')
```

## 8. Phase-velocity polar plots at points on the CMA diagram

`phase_velocity_polar(X, Y, mu)` is the static equivalent of clicking a point: it plots $v_{ph}/c=1/n$ versus $\theta$ for both roots ($0°$ = along $\mathbf B_0$, clockwise like CMA-Lab), labels the roots by their $\theta=0$ limit (R/L) and $\theta=90°$ limit (O/X), marks the resonance cone $\tan^2\theta_{res}=-P/S$ if it exists, and draws the vacuum circle $v_{ph}=c$ (dashed).

Wave-normal surface types (Bellan Fig. 6.3): **ellipsoid** (propagates at all $\theta$), **dumbbell** (only for $\theta<\theta_{res}$), **wheel** (only for $\theta>\theta_{res}$).

```python
def phase_velocity_polar(X, Y, mu=MU_TEACH, ax=None, title=None):
    p = cp.stix(X, Y, mu)
    th = np.linspace(0, 2*np.pi, 1441)
    vp, vm = cp.phase_velocity(th, p['S'], p['P'], p['R'], p['L'])
    names = cp.label_modes(p['S'], p['P'], p['R'], p['L'])
    if ax is None:
        fig = plt.figure(figsize=(5, 5)); ax = fig.add_subplot(projection='polar')
    ax.set_theta_zero_location('N'); ax.set_theta_direction(-1)
    vmax = np.nanmax(np.concatenate([vp, vm, [0]]))
    info = []
    for v, colr, nm in [(vp, '#2a9d8f', names[0]), (vm, '#c9a227', names[1])]:
        if not np.isfinite(v).any():
            continue
        typ = cp.wave_normal_type(th, v)
        ax.plot(th, v, color=colr, lw=3, label=f"{'/'.join(x for x in nm if x) or '?'}: {typ}")
        info.append((nm, typ))
        if nm[0]:
            ax.text(0, v[0] * 1.08, nm[0], color='#50258c', ha='center', fontsize=11)
        i90 = len(th) // 4
        if nm[1] and np.isfinite(v[i90]):
            ax.text(np.pi/2, v[i90] * 1.12, nm[1], color='#50258c', va='center', fontsize=11)
    if vmax > 0 and vmax < 4:
        ax.plot(th, np.ones_like(th), 'k--', lw=0.8)       # v_ph = c
    if -p['P'] / p['S'] > 0:
        tres = np.arctan(np.sqrt(-p['P'] / p['S']))
        for t in (tres, np.pi - tres, np.pi + tres, 2*np.pi - tres):
            ax.plot([t, t], [0, max(vmax, 1e-3) * 1.15], color='crimson', lw=0.8, ls='-.')
    ax.set_rmax(max(vmax, 1e-3) * 1.25)
    ax.set_title(title or f'$v_{{ph}}/c$  (X={X:.3g}, Y={Y:.3g}, μ={mu:g})', fontsize=10)
    if info:
        ax.legend(loc='lower center', bbox_to_anchor=(0.5, -0.32), fontsize=8, frameon=False)
    else:
        ax.text(0, 0, 'no propagating mode', ha='center', fontsize=10)
    return p, info

# two example points, like the two clicks in the CMA-Lab screenshot
fig = plt.figure(figsize=(10, 5))
for i, (X, Y) in enumerate([(0.2, 0.6), (0.2, 1.6)], 1):
    ax = fig.add_subplot(1, 2, i, projection='polar')
    p, info = phase_velocity_polar(X, Y, ax=ax, title=f'({i})  X={X}, Y={Y}')
    print(i, {k: round(float(v), 3) for k, v in p.items()}, info)
fig.tight_layout(); fig.savefig(FIG / 'polar_two_points.png', dpi=150, bbox_inches='tight')
```

## 10. Saturn and Jupiter: where Cassini and Juno events sit on the CMA diagram

### 10.1 Axis choice: $\omega_{pe}^2/\omega^2$ or $(\omega_{pe}^2+\omega_{pi}^2)/\omega^2$?
Both are fine. For electrons plus one ion species, $\omega_{pi}^2=\omega_{pe}^2/\mu$, so Bellan's axis is $X(1+1/\mu)$: the same picture with the horizontal axis stretched by a constant factor ($1.4$ for $\mu=2.5$, $1.0005$ for H⁺, $1.00003$ for water ions). Region shapes and the modes in each region are unchanged. Only the label of the $P=0$ line moves: $X=\mu/(\mu+1)$ on our axis, exactly 1 on Bellan's. For several ion species (H⁺ + W⁺ at Saturn; O⁺, S⁺, S²⁺… at Jupiter), $\omega_{pe}^2/\omega^2$ stays a clean density variable, while the sum $\sum\omega_{ps}^2$ also depends on composition. Bellan also squares the vertical axis ($\Omega_e^2/\omega^2$); that changes the curve shapes but not the regions.

### 10.2 The fixed-plasma ray
At one location $n_e$ and $|B|$ are fixed, so $\omega_{pe}$ and $\Omega_e$ are fixed and only the wave frequency $\omega$ is free. Then $Y^2/X=\Omega_e^2/\omega_{pe}^2$ is constant, i.e. the location is the curve $Y=(\Omega_e/\omega_{pe})\sqrt X$. High frequency is near the origin; lowering $\omega$ moves the point outward along the curve, through the electron features ($\omega_{pe}$, $\Omega_e$), the lower hybrid, and finally the ion cyclotron resonance $Y=\mu$ and the Alfvén region beyond it. On Bellan's log axes ($X$ vs $Y^2$) this curve is a straight line of slope 1, hence "ray". Reading along the ray tells you which modes can exist at that location at each frequency. The one number that sets the ray is $\Omega_e/\omega_{pe}$ (≈ 0.01–0.3 in the equatorial magnetospheres below, ≫ 1 in Jupiter's polar regions).

### 10.3 Event points
For an observed wave frequency $f$: $Y=f_{ce}/f$ and $X=(f_{pe}/f)^2$, with $f_{ce}=27.99\,|B|_{nT}$ Hz and $f_{pe}=8.979\,\sqrt{n_e[\mathrm{cm^{-3}}]}$ kHz. The mass ratio is the ICW ion used by the quick-check notebooks. A doubly charged ion of mass $m$ acts exactly like a singly charged ion of mass $m/2$ in the cold-plasma formulas ($\omega_{pi}^2$ and $\Omega_i$ both scale as $Z/m$ at fixed $n_e$), so O²⁺ uses $\mu = 8\,m_u/m_e$.

$|B|$ is the mean measured field in each window (read from the MAG files used by `cassini_quick_check.ipynb` and `jupiter/juno_quick_check.ipynb`). Juno positions come from the same files. $n_e$ is **not measured here**; it is a typical model value:
- Saturn: Kamran et al. (2025) TRAPPED-S, Figs. 6–8 (RPWS upper-hybrid electron density inside 8 $R_S$; CAPS-FITI W⁺ + H⁺ outside). Peak ≈ 60–70 cm⁻³ near 4–5 $R_S$, consistent with Rusaitis et al. (2021).
- Jupiter: typical literature values (Bagenal & Delamere 2011; Bagenal et al. 2015; Bagenal & Dols 2020). Not checked against Juno Waves densities for these passes.

Treat $n_e$ as uncertain by a factor of ~2. $X$ scales with $n_e$, and the mode types below do not change over that range.

```python
import pandas as pd

AMU_ME = 1822.888   # atomic mass unit / electron mass
# |B| measured (window mean); n_e typical model value (see text); A = ion mass in amu, Z = charge; f = observed peak (Hz)
EVENTS = [
    dict(name='Saturn L≈4.7, ICW (Meeks+2016)',      utc='2010-12-21 03:00–03:10', B=185.8, ne=60,   A=18, Z=1, ion='H2O+', f=0.1433),
    dict(name='Saturn L≈6.6, ICW+mirror (Chou&Cheng 2017)', utc='2009-10-13 20:50–21:00', B=58.6, ne=20, A=18, Z=1, ion='H2O+', f=0.040),
    dict(name='Saturn L≈7.1, ICW (Chou&Cheng 2017)', utc='2007-09-30 02:42–02:52', B=49.2,  ne=15,   A=18, Z=1, ion='H2O+', f=0.039),
    dict(name='Jupiter Io torus 5.9 RJ, Juno PJ57',  utc='2023-12-30 08:14–08:58', B=2000., ne=2000, A=32, Z=1, ion='S+',   f=0.534),
    dict(name='Jupiter Europa orbit 9.4 RJ, Juno PJ45', utc='2022-09-29 09:20–09:50', B=436., ne=150, A=16, Z=2, ion='O++', f=0.458),
    dict(name='Jupiter 93 RJ pre-dawn, EMIC (Chakraborty+2025)', utc='2017-07-24 01:49–01:57', B=7.2, ne=0.05, A=16, Z=2, ion='O++', f=0.0083),
]

def event_point(B, ne, A, Z, f, **_):
    fce, fpe = 27.9925 * B, 8978.66 * np.sqrt(ne)
    mu = A / Z * AMU_ME
    return dict(fce=fce, fpe=fpe, mu=mu, fci=fce / mu, ratio=fce / fpe, X=(fpe / f) ** 2, Y=fce / f)

rows = []
th = np.linspace(0, 2 * np.pi, 1441)
for ev in EVENTS:
    e = event_point(**ev)
    p = cp.stix(e['X'], e['Y'], e['mu'])
    vs = cp.phase_velocity(th, p['S'], p['P'], p['R'], p['L'])
    modes = []
    for nm, v in zip(cp.label_modes(p['S'], p['P'], p['R'], p['L']), vs):
        typ = cp.wave_normal_type(th, v)
        if typ != 'none':
            modes.append(f"{'/'.join(x for x in nm if x)} {typ}, v_ph(0)/c={v[0]:.1e}")
    rows.append({'event': ev['name'], 'UTC': ev['utc'], '|B| nT': ev['B'], 'n_e cm⁻³': ev['ne'], 'ion': ev['ion'],
                 'f Hz': ev['f'], 'f/f_ci': round(ev['f'] / e['fci'], 2), 'Ωe/ωpe': round(e['ratio'], 3),
                 'mi/me': round(e['mu'], 1), 'X': f"{e['X']:.3g}", 'Y': f"{e['Y']:.4g}",
                 'θres °': round(float(np.degrees(np.arctan(np.sqrt(-p['P'] / p['S'])))), 2) if -p['P'] / p['S'] > 0 else None,
                 'modes': '; '.join(modes)})
pd.set_option('display.max_colwidth', None)
pd.DataFrame(rows)
```

**What the table shows.** Every event point lies on the far side of the ion cyclotron resonance ($Y>\mu$, i.e. $f<f_{ci}$), deep in the Alfvén region of the CMA diagram ($X\sim10^{8}$–$10^{12}$). Two modes exist there:
- the **L-mode (shear Alfvén / ion cyclotron) wave**, a dumbbell confined inside the resonance cone ($\theta_{res}\approx89.2°$–$89.8°$, so it looks almost like $v_{ph}\approx v_A\cos\theta$), whose $v_{ph}\to0$ as $f\to f_{ci}$ along $\mathbf B_0$. This is the left-handed, nearly field-aligned wave that the ICW classifier looks for.
- the **R/X fast (compressional Alfvén / magnetosonic) wave**, a near-sphere with $v_{ph}\approx v_A$.

$v_{ph}/c\sim10^{-4}$–$10^{-2}$, i.e. $v_{ph}\approx v_A$. Open `cma_interactive.html` and choose a **Location** preset to see each point, its ray and its polar plot.

Limits of this picture. (1) Real Saturn and Jupiter plasmas have several ion species; with H⁺ + W⁺ there are extra cutoffs and crossover frequencies between the two gyrofrequencies that a one-ion model cannot show. (2) Near $f\approx f_{ci}$ thermal effects (cyclotron damping, finite Larmor radius) matter. (3) **Mirror modes are not cold-plasma waves.** They are non-propagating structures driven by $T_\perp>T_\parallel$, so they need a warm (kinetic or anisotropic-MHD) description, even though their frequency falls in the same Alfvén region.
