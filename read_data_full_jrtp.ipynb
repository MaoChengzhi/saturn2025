## find the file - time index

because all the data in 10 years are stored in different files, in a total of 300GB.


```python
tab_files=glob.glob('data/_cassini-mag-cal_full_krtp/*.TAB')
len(tab_files)
```

```python
file_name_list=[]
start_time_list=[]
end_time_list=[]

for tab_file in tab_files:
    # columns TIME,BR,BTHETA,BPHI,BTOTAL,NPTS
    temp_df=pd.read_csv(tab_file,sep=r'\s+',names=['TIME','BR','BTHETA','BPHI','BTOTAL','NPTS'])
    
    # I do not quite understand why sep='\s+' does not work,  
    
    # Why '\s+' warns in Python
    # '\s+' is a normal Python string literal.
    # In normal strings, backslash starts a Python escape (\n, \t, \\, etc.).

    # \s is not a valid Python escape sequence.
    # So Python emits: SyntaxWarning: invalid escape sequence '\s'.
    # Regex would understand \s, but Python complains before regex sees it.


    start_time=temp_df['TIME'].iloc[0]
    end_time=temp_df['TIME'].iloc[-1]
    
    # verify if the start time is the minimum of temp_df['TIME'], and end time is the latest of temp_df
    try: 
        assert start_time==temp_df['TIME'].min()
        assert end_time==temp_df['TIME'].max()
    except:
        # raise error
        raise ValueError(f"Error in file {tab_file}: start_time or end_time mismatch")
    
    file_name_list.append(tab_file)
    start_time_list.append(start_time)
    end_time_list.append(end_time)

```

```python
index_df=pd.DataFrame({'file_name':file_name_list,
                          'start_time':start_time_list,
                            'end_time':end_time_list})
```

```python
with open('data/full_krtp_file_time_index.csv','wb') as f:
    pickle.dump(index_df,f)
```
